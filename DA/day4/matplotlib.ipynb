```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y = np.sin(x)
print(x)
print(y)
 

plt.plot(x, y)
```

### Sine wave

```python
plt.figure(figsize=(10, 2))
plt.title("Sine Wave")
plt.xlabel("x")
plt.ylabel("y")
plt.plot(x, y, color="blue", linestyle="", linewidth=2,
         marker="o", markersize=4, markerfacecolor="red", markeredgecolor="black",
         label="sin(x)")
plt.legend()
plt.xticks(np.arange(0, 11, 1))
plt.yticks(np.arange(-1, 1.1, 0.2))
plt.grid()
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(2*x)
plt.figure(figsize=(10, 2))
plt.plot(x, y1, color="blue", linestyle="-", linewidth=2, label="sin(x)")
plt.show()
plt.figure(figsize=(10, 2))
plt.plot(x, y2, color="red", linestyle="-", linewidth=2, label="cos(2x)")
plt.show()
```

### Bar Plot

```python
students = ["Alice", "Bob", "Charlie", "David"]
scores = [85, 90, 78, 92]
plt.figure(figsize=(8, 2))
plt.bar(students, scores, color="green")
plt.title("Student Scores")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.ylim(0, 100)
plt.show()
```

```python
n = 10_000_000
dice = np.random.randint(1, 7, size=(n, 2))
# print(dice)
dice_d = dice[dice[:, 0] == dice[:, 1]]
dice_n = dice[dice[:, 0] != dice[:, 1]]
# Not Doubles
# print(dice_d)
# print(dice_n)
opt_f_n = dice_n.flatten()
# print(opt_f_n)
opt_f_sums_n = dice_n.sum(axis=1)
# print(opt_f_sums_n)
opt_n = np.concat([opt_f_n, opt_f_sums_n])
# print(opt_n)
 
# Doubles
opt_d = (dice_d[:, :1] @ np.array([[1, 2, 3, 4]])).flatten()
# print(opt_d)
opt = np.concat([opt_n, opt_d])
# print(opt)
 
# agg and display the results
values, counts = np.unique(opt, return_counts=True)
probs = counts / n
plt.figure(figsize=(11, 2))
plt.bar(values, probs, color="blue")
plt.title("Probability Distribution of Backgammon Dice Rolls")
plt.xlabel("Distance")
plt.ylabel("Probability")
plt.xticks(values)
for v, p in zip(values, probs):
    plt.text(v, p + 0.005, f"{p:.2f}", ha="center", va="bottom")
plt.ylim(0, max(probs) + 0.2)
plt.show()
```

### Scatter Plot

```python
x = np.arange(10)
y = np.random.rand(10)
size = np.random.rand(10) * 1000  # Bubble sizes
colors = np.random.rand(10)  # Bubble colors
plt.figure(figsize=(8, 2))
plt.scatter(x, y, s=size, c=colors, cmap="hot")
plt.colorbar()
plt.show()
 
```

### Histogram

```python
x = np.random.randn(1000)
plt.figure(figsize=(10, 2))
counts, vals, patches = plt.hist(x, bins=10, color='skyblue', edgecolor='black')
plt.xticks(vals);
plt.bar_label(patches, labels=[int(c) for c in counts], padding=3)
plt.ylim(0, max(counts)*1.2)
 
```

### Hist2d

### multiple plots

```python
x = np.linspace(0, 10, 1000)

y1 = np.sin(x)
y2 = np.sin(2*x)
y3 = np.sin(3*x)
y4 = np.cos(x)
y5 = np.cos(2*x)
y6 = np.cos(3*x)

fig, ax = plt.subplots(2, 3, figsize=(10, 4))
ax[0,0].plot(x, y1)
ax[0,0].set_title("sin(x)")
ax[0,1].plot(x, y2)
ax[0,1].set_title("sin(2x)")
ax[0,2].plot(x, y3)
ax[0,2].set_title("sin(3x)")
ax[1,0].plot(x, y4)
ax[1,0].set_title("cos(x)")
ax[1,1].plot(x, y5)
ax[1,1].set_title("cos(2x)")
ax[1,2].plot(x, y6)
ax[1,2].set_title("cos(3x)")
fig.tight_layout()
plt.show()
```
