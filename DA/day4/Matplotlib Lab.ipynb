Create a function:

    def plot(fn,a,b,num_points)

Shows the plot of function fn between a and b using num_points


```python
def plot(fn, a, b, num_points):
    x = np.linspace(a, b, num_points)
    y = fn(x)
    plt.plot(x, y)
    plt.show()

plot(np.sin, 0, 2 * np.pi, 100)
```

```python
def fn(x):
    return x * np.cos(1 / x)

plot(fn, 0 ,np.pi/4, 1000)
plot(fn, -0.1, 0.1, 1000)
plot(fn, 0, 2, 1000)
```
