### Groupby

```python
df_gb = df.groupby('day')[['total_bill', 'tip']].mean()
display(
    "df_gb.loc['Sun', 'tip']:", df_gb.loc['Sun', 'tip'],
    "df_gb.loc['Sun']:", df_gb.loc['Sun'],
    "df_gb.loc['Sun', :]:", df_gb.loc['Sun', :],
    "df_gb.loc[':Fri', ['tip']]:", df_gb.loc[:'Fri', ['tip']]
)
```

```python
def count_of_smokers(x):
    return x[x == 'Yes'].count()
 
df.groupby('day').agg(
    total_bill_sum=('total_bill', 'sum'),
    total_bill_max=('total_bill', np.max),
    tip_mean=('tip', 'mean'),
    smokers_count=('smoker', count_of_smokers)
)
```

```python
# EX1:
# I want to know what is the best shift (day and time) to work, I need to know:
# 1. what is the sum and mean of the tip
# 2. what is the median of the total bill
# 3. what is the percantage of smokers
# use groupby
 
def smk_per(x):
    return x[x == 'Yes'].count()/x.count()
 
df.groupby(['day', 'time']).agg(
    tip_sum=('tip', 'sum'),
    tip_mean=('tip', 'mean'),
    total_bill_median=('total_bill', 'median'),
    smokers_percent=('smoker', smk_per)
).sort_values(by='tip_sum', ascending=False)
```

```python
total_paid = df['total_bill'] + df['tip']
 
df.assign(total_paid=total_paid).head(2)
```

```python
df.head(2)
```

```python
df.assign(total_paid=total_paid).groupby(['day', 'time']).agg(
    total_paid_sum=('total_paid', 'sum'),
    tip_sum=('tip', 'sum'),
    tip_mean=('tip', 'mean'),
    total_bill_median=('total_bill', 'median'),
    smokers_percent=('smoker', smk_per),
    smokers_count=('smoker', count_of_smokers)
).sort_values(by='tip_sum', ascending=False)
```

```python
(
    df
    .assign(total_paid=df['total_bill'] + df['tip'])
    .groupby(['day', 'time'])
    .agg(
        total_paid_sum=('total_paid', 'sum'),
        tip_sum=('tip', 'sum'),
        tip_mean=('tip', 'mean'),
        total_bill_median=('total_bill', 'median'),
        smokers_percent=('smoker', smk_per),
        smokers_count=('smoker', count_of_smokers)
    )
    .sort_values(by='tip_sum', ascending=False)
)
```
