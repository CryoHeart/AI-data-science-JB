Write a function that takes as input a 2-D ndarray and scales the last row and column by 2

```python
def scale_last_row_and_col_by_2(nparray):
    nparray[-1, :] *= 2
    nparray[:, -1] *= 2
    return nparray

arr = np.random.rand(5, 5)
scaled_arr = scale_last_row_and_col_by_2(arr)
print("Original array:\n", arr)
print("Scaled array:\n", scaled_arr)
```

Compute the dot product of arrays [1,2,3,4] and [cos(1),cos(2), cos(3),cos(4)], but first you have to reshape them into  2-dimensional arrays


```python
arr1 = [1,2,3,4]
arr2 = [np.cos(1), np.cos(2), np.cos(3), np.cos(4)]
arr1 = np.array(arr1).reshape(2, 2)
arr2 = np.array(arr2).reshape(2, 2)
result = np.dot(arr1, arr2)
result
```
