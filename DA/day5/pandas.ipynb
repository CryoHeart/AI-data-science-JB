### Series operations

```python
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([10, 20, 30, 40, 50])

print("s1+s2 = \n", s1 + s2)
print("s1-s2 = \n", s1 - s2)
print("s1*s2 = \n", s1 * s2)
print("s1 * 2 = \n", s1 * 2)
```

```python
print("s1 == s2 = \n", s1 == s2)
```

### Series Aggregations

```python
s1 = pd.Series([1, 2, 3, 4, 5])
print("s1.sum() = ", s1.sum())
print("s1.mean() = ", s1.mean())
print("s1.max() = ", s1.max())
print("s1.count() = ", s1.count())
```

```python
s1 = pd.Series([1, 2, 2, 1, 5])
s1.value_counts()
```

### Handeling missing data in series

```python
s1 = pd.Series(index=['a', 'b', 'c', 'd', 'e'], data=[1, 2, 3, 4, 5])
s2 = pd.Series(index=['a', 'b', 'd', 'e'], data=[10, 20, 30, 40])
print("s1:\n", s1)
print("s2:\n", s2)
```

```python
s3 = s1 + s2
print("s3:\n", s3)
```

```python
s3.dropna()
```

```python
s3.isna()
```

### Filtering data in series

```python
s3[s3.isna()]
```

```python
s3[s3 > 25]
```
