```python
import matplotlib.pyplot as plt
import numpy as np
```

### 3d

```python
fig, ax = plt.subplots(subplot_kw={'projection': '3d'})
```

### 3d scatter plot

```python
x = np.random.randn(10000)
y = np.random.randn(10000)
z = np.random.randn(10000)
color = np.random.choice(['red', 'blue'], size=10000)


fig, ax = plt.subplots(subplot_kw={'projection': '3d'})
ax.scatter(x, y, z, c=color, alpha=0.1)
# ax.set_xlim3d(-1, 1)
# ax.set_ylim3d(-1, 1)
# ax.set_zlim3d(-1, 1)
plt.show()
```

### surface plot

```python
x = np.array([
    [1, 2, 3],
    [1, 2, 3],
    [1, 2, 3],
])
y = np.array([
    [1, 1, 1],
    [2, 2, 2],
    [3, 3, 3],
])
z = x + y

fig, ax = plt.subplots(subplot_kw={'projection': '3d'})
ax.plot_surface(x, y, z, cmap='viridis', edgecolor='black', alpha=0.5)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
# roteate the plot
ax.view_init(elev=0, azim=30, roll=0)
# plt.show()
```
