```python
import numpy as np
import matplotlib.pyplot as plt
```

Create a numpy array with 100 numbers from 1 to 20. You are not allowed to use any loop

```python
x = np.linspace(1, 20, 100)
```

Create a numpy array containing the sine of the numbers from the above array. 

```python
y = np.sin(x)
```

Create a graph with the last 2 arrays


```python
plt.figure(figsize=(10, 2))
plt.plot(x, y);
```

```python
def plot(fn, a, b, num_points=100):
    x = np.linspace(a, b, num_points)
    y = fn(x)
    plt.figure(figsize=(10, 2))
    plt.plot(x, y, color='blue', linewidth=2)
    plt.xlabel('x')
    plt.ylabel('y')
    plt.grid(True)
    plt.show()

plot(np.sin, 0, 2 * np.pi, 100)
```

```python
def cool_fn(x):
    return x * np.cos(1/x)

plot(cool_fn, -0.1, 0.1, 1000)
```

Create a text file with 2 lines of numbers:

23 34 45 35 60 40 56 58<br>
2 5 6 8 12 15 18 25 

load the file into numpy array

```python
data = np.loadtxt('data.txt')
```

Create a solid line graph from the above data (first line is y)

```python
plt.figure(figsize=(10, 2))
plt.plot(data[1, :], data[0, :], color='green', linewidth=2);
```

Change the graph to green solid line with square markers


```python
plt.figure(figsize=(10, 2))
plt.plot(data[1, :], data[0, :], color='green', linewidth=2, marker='s');
```

Change the graph to pie graph(use only the first line) 


```python
plt.figure(figsize=(4, 4))
plt.pie(data[0, :], labels=data[1, :], autopct='%1.1f%%', startangle=90)
plt.legend(title='Categories', loc=(1.2,0.3));
```
