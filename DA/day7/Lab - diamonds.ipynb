```python
fig, axes = plt.subplots(2, 4, figsize=(15, 4))

clarity_sorted = ['IF', 'VVS1', 'VVS2', 'VS1', 'VS2', 'SI1', 'SI2', 'I1']
for i, clarity in enumerate(clarity_sorted):
    ax = axes[i // 4, i % 4]
    df_to_show = df[df['clarity'] == clarity]
    ax.hist2d(df_to_show['price'], df_to_show['carat'], bins=30, cmap='hot')
    ax.set_title(f'Clarity: {clarity}')
    ax.set_xlabel('Price')
    ax.set_ylabel('Carat')
    ax.set_xlim(0, df_to_show['price'].max())
    ax.set_ylim(0, df_to_show['carat'].max())

fig.tight_layout()
```

```python
fig, axes = plt.subplots(2, 4, figsize=(15, 4))
 
clarity_sorted = ['IF', 'VVS1', 'VVS2', 'VS1', 'VS2', 'SI1', 'SI2', 'I1']
for i, clarity in enumerate(clarity_sorted):
    ax = axes[i // 4, i % 4]
    df_to_show = df[df['clarity'] == clarity]
    _, _, _, im = ax.hist2d(df_to_show['price'], df_to_show['carat'], bins=30, cmap='hot')
    ax.set_title(f'Clarity: {clarity}')
    ax.set_xlabel('Price')
    ax.set_ylabel('Carat')
    ax.set_facecolor('black')
    plt.colorbar(im, ax=ax, orientation='vertical')
    
fig.tight_layout()
 
```

```python
import numpy as np
# twinx shares the x-axis and gives each side its own y-axis scale.
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.exp(x)
fig, ax1 = plt.subplots(figsize=(8, 5))
ax2 = ax1.twinx()
 
ax1.plot(x, y1, 'b-', label='sin(x)')
ax1.set_xlabel('x')
ax1.set_ylabel('sin(x)', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')
 
ax2.plot(x, y2, 'r-', label='exp(x)')
ax2.set_ylabel('exp(x)', color='red')
ax2.tick_params(axis='y', labelcolor='red')
fig.tight_layout()
 
 
```

# Data preparing 

delete all the diamonds with more than 10 percent difference between x and y

```python
diff_pct = (df['x'] - df['y']).abs() / df[['x', 'y']].max(axis=1)
df = df[diff_pct <= 0.10]
sns.scatterplot(data=df, x='x', y='y')
plt.show()
```

delete all the diamonds with z>x

```python
df = df[df['z'] <= df['x']]
sns.scatterplot(data=df, x='x', y='z')
plt.show()
```

Create a new column 'xy' with the mean value of x and y

```python
df['xy'] = df[['x', 'y']].mean(axis=1)
```

```python
def multi_hist(col1, col2, ax1=None):
    if ax1 is None:
        fig, ax1  = plt.subplots(figsize=(8, 5))
    ax2 = ax1.twinx()
 
    # y= makes the histogram horizontal; kde=True scales the curve to counts.
    sns.histplot(y=col1, bins=10, stat='count', kde=True, alpha=0.4,
                 color='blue', label=col1.name, ax=ax1)
    # sns.kdeplot(y=col1, ax=ax1, color='blue', label=col1.name, alpha=0.4)
    ax1.set_xlabel('Count')
    ax1.set_ylabel(col1.name, color='blue')
    ax1.tick_params(axis='y', labelcolor='blue')
 
    sns.histplot(y=col2, bins=10, stat='count', kde=True, alpha=0.4,
                 color='green', label=col2.name, ax=ax2)
    # sns.kdeplot(y=col2, ax=ax2, color='green', label=col2.name, alpha=0.4)
    ax2.set_ylabel(col2.name, color='green')
    ax2.tick_params(axis='y', labelcolor='green')
 
fig, axes = plt.subplots(2, 4, figsize=(15, 10))
 
clarity_sorted = ['IF', 'VVS1', 'VVS2', 'VS1', 'VS2', 'SI1', 'SI2', 'I1']
for i, clarity in enumerate(clarity_sorted):
    ax = axes[i // 4, i % 4]
    df_to_show = df[df['clarity'] == clarity]
    multi_hist(df_to_show['carat'], df_to_show['price'], ax1=ax)
    ax.set_title(f'clarity:{clarity} cnt:{len(df_to_show)}')
    
fig.tight_layout()
 
```
