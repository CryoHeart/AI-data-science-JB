```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

# Seaborn

```python
df = sns.load_dataset('tips')
df.head()
```

### Displot

```python
sns.displot(df['total_bill'], bins=30, kde=True, aspect=5)
```

### Joint Plot

```python
sns.jointplot(x='total_bill', kind='scatter', hue='sex', y='tip', data=df, height=4)
```

```python
sns.displot(data=df, x="total_bill", bins=30, kde=True, aspect=5, hue="sex")
```

```python
sns.pairplot(data=df, hue="sex", height=1)
```

```python
sns.pairplot(data=df, hue="sex", height=2, kind='kde', plot_kws=dict(alpha=0.5, fill=True));
 
```

```python
plt.figure(figsize=(10, 2))
sns.countplot(data=df, x="day", hue="sex");
 
```

### Cound plot

```python
fig, ax = plt.subplots(figsize=(10, 2))
sns.countplot(data=df, x="day", hue="sex", ax=ax);
```

### Box plot

```python
plt.figure(figsize=(10, 3))
sns.boxplot(data=df)
```
