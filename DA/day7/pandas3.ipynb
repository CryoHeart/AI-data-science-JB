```python
import pandas as pd
import seaborn as sns
import numpy as np
```

### Concatinating data

```python
df1 = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6],
    'C': [7, 8, 9]
})

df2 = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6],
    'C': [7, 8, 9]
})

df3 = pd.concat([df1, df2])

df3.reset_index(drop=True)
```

### Joining data

```python
df1 = pd.DataFrame({
    'id': [1, 2, 3, 4, 5],
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'age': [25, 30, 35, 40, 45]
})

df2 = pd.DataFrame({
    'id': [1, 2, 3, 4, 6],
    'salary': [5000, 6000, 7000, 8000, 9000],
    'position': ['Manager', 'Developer', 'Designer', 'Tester', 'HR']
})

display(df1, df2)
```

```python
pd.merge(left=df1, right=df2, on='id', how='inner')
```

```python
pd.merge(left=df1, right=df2, on='id', how='outer')
```

```python
pd.merge(left=df1, right=df2, on='id', how='right')
```
