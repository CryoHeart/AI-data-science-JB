```python
def do_math(a,b, operation="add"):
    if operation == "add":
        return a + b
    elif operation == "sub":
        return a - b
    elif operation == "mul":
        return a * b
    elif operation == "div":
        return a / b
    else:
        raise ValueError("Invalid operation")

print(do_math(1, 2))
print(do_math(1, 2, operation="sub"))
print(do_math(1, 2, operation="mul"))
print(do_math(1, 2, operation="div"))
```

```python
def my_agg(*nums, op="sum"):
    print("type(nums):", type(nums), "nums:", nums, "op:", op)
    if op == "sum":
        total = 0
        for num in nums:
            total += num
        return total
    elif op == "prod":
        total = 1
        for num in nums:
            total *= num
        return total
 
print(my_agg(1, 2, 3.5)) # default op="sum"
print(my_agg(1, 2, 3, 4, 5, 6, 7, op="prod"))
 
```

### Closure

```python
def create_gun_with_ammo(bullet_type, ammo):
    def shoot(sound):
        nonlocal ammo
        if ammo > 0:
            ammo -= 1
            print(f"Bang! {bullet_type} {sound}. Ammo left: {ammo}")
        else:
            print("Click! RELOAD!")
    return shoot

anti_superman_gun = create_gun_with_ammo("Kryptonite", 3)
anti_vampire_gun = create_gun_with_ammo("Silver", 5)
pikachu_gun = create_gun_with_ammo("Thunderbolt", 2)
anti_superman_gun("bang!")
anti_superman_gun("Pow!")
anti_superman_gun("Tak!")
anti_vampire_gun("Boom!")
anti_vampire_gun("Brrr!")
anti_vampire_gun("BOING!")
pikachu_gun("Pika... CHOOOO!")
```

### Decorators

```python
def print_and_wrap_with_stars(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        result = f"* {result} *"
        stars = "*" * len(result)
        print(f"{stars}\n{result}\n{stars}")
    return wrapper
 
@print_and_wrap_with_stars
def greet(name):
    return f"Hello, {name}!"
 
@print_and_wrap_with_stars
def bye(name):
    return f"Goodbye, {name}!"
 
@print_and_wrap_with_stars
def greet_students(kita, students):
    greeting = kita + ": "
    for student in students:
        greeting += f"Hello, {student}! | "
    return greeting
 
greet("Alice")
bye("Bob")
bye("Charlie")
greet_students("Alef1", ["Alice", "Bob", "Charlie"])
```

```python
# Challenge: Implement a configurable decorator that for print_and_wrap_with_stars decorator that
# allows the user to specify the character to use for wrapping the output. For example, if the user specifies
# the character "*", the output should be wrapped with asterisks.
# @print_and_wrap_with_stars('*')
# def greet(name):
#     return f"Hello, {name}!"
 
# @print_and_wrap_with_stars('#')
# def bye(name):
#     return f"Goodbye, {name}!"
# greet("Alice")
# bye("Bob")
# output:
# *****************
# * Hello, Alice! *
# *****************
# #################
# # Goodbye, Bob! #
# #################

def print_and_wrap(char="*"):
    def decorator(func):
        def wrapper(*args, **kwargs):
            result = func(*args, **kwargs)
            wrapped = f"{char} {result} {char}"
            border = char * len(wrapped)
            print(border)
            print(wrapped)
            print(border)
        return wrapper
    return decorator

@print_and_wrap("*")
def greet(name):
    return f"Hello, {name}!"

@print_and_wrap("#")
def bye(name):
    return f"Goodbye, {name}!"

greet("Alice")
bye("Bob")
```
