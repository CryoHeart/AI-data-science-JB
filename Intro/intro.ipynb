### Jupyter Cheatsheet
 
**Cell States:**
- command mode: press `Esc` to enter command mode
- edit mode: press `Enter` to enter edit mode
 
**Running Cells:**
- `Shift + Enter`: run the current cell and move to the next cell
- `Ctrl + Enter`: run the current cell and stay in the same cell
- `Alt + Enter`: run the current cell and insert a new cell below
 
**Keyboard Shortcuts (in command mode):**
- `A`: insert a new cell above
- `B`: insert a new cell below
- `X`: cut the selected cell
- `C`: copy the selected cell
- `V`: paste the cell below
- `shift + L`: toggle line numbers
- `M`: change the cell type to Markdown
- `Y`: change the cell type to Code

# List

```python
nums = [1, 2, 3, 4, 5]
print("nums:", nums, "type(nums):", type(nums))
print("nums[0]:", nums[0])
print("nums[-1]:", nums[-1])
nums.append(6)
print("nums.append(6):", nums)
nums.insert(0, 0)
print("nums.insert(0, 0):", nums)
nums.insert(3, 2.5)
print("nums.insert(3, 2.5):", nums)
# slices are references to the original list, not copies
nums[2:4] = [20, 30]
print("nums[2:4] = [20, 30]:", nums)
nums.remove(30) # removes the first occurrence of 30
print("nums.remove(30):", nums)
item = nums.pop() # removes and returns the last item
print("nums.pop():", item, "nums:", nums)
item = nums.pop(2) # removes and returns the item at index 2
print("nums.pop(2):", item, "nums:", nums)
```
