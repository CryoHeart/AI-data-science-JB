### Input Validation

```python
class Car:
    def __init__(self, consumption, fuel_capacity):
        self.__consumption = consumption  # liters per 100 km
        self.__fuel_capacity = fuel_capacity  # liters
        self.__fuel_level = fuel_capacity  # start with a full tank
        self.__odometer = 0  # kilometers driven
 
    @property
    def consumption(self):
        return self.__consumption
 
    @property
    def fuel_capacity(self):
        return self.__fuel_capacity
 
    @property
    def fuel_level(self):
        return self.__fuel_level
 
    @fuel_level.setter
    def fuel_level(self, value):
        if 0 <= value <= self.__fuel_capacity:
            self.__fuel_level = value
        else:
            raise ValueError(f"Fuel level must be between 0 and {self.__fuel_capacity} liters.")
 
    @property
    def odometer(self):
        return self.__odometer
 
    def drive(self, distance):
        fuel_needed = (self.__consumption / 100) * distance
        if fuel_needed <= self.__fuel_level:
            self.__fuel_level -= fuel_needed
            self.__odometer += distance
            return True
        else:
            # Drive as far as possible with the remaining fuel
            max_distance = (self.__fuel_level / self.__consumption) * 100
            self.__odometer += max_distance
            self.__fuel_level = 0
            return False
 
 
c1 = Car(consumption=8, fuel_capacity=50)
print(f"Fuel level: {c1.fuel_level} liters, odometer: {c1.odometer} km")
c1.drive(100)
print(f"Fuel level: {c1.fuel_level} liters, odometer: {c1.odometer} km")
c1.fuel_level = 50
print(f"Fuel level: {c1.fuel_level} liters, odometer: {c1.odometer} km")
# c1.fuel_level = 60  # This will raise a ValueError
```

```python
# EX1: refactor the Point2d and Rectangle classes to use properties for x, y, w, h, and p.
# Ensure that the properties enforce that w and h are always positive numbers.
# If a negative value is assigned to w or h, raise a ValueError.
 
class Point2d:
    def __init__(self, x, y):
        self.__x = x
        self.__y = y
 
    @property
    def x(self):
        return self.__x
 
    @property
    def y(self):
        return self.__y
 
    def __repr__(self):
        return f"p2d({self.__x},{self.__y})"
 
    def __add__(self, other):
        if isinstance(other, Point2d):
            # return Point2d(self.__x + other.__x, self.__y + other.__y)
            self.__x += other.__x
            self.__y += other.__y
            return self
        return NotImplemented
 
    def translate(self, dx, dy):
        self.__x += dx
        self.__y += dy
 
class Rectangle:
    def __init__(self, p, w, h):
        self.__p = p  
        self.__w = w  
        self.__h = h
 
    @property
    def p(self):
        return self.__p
 
    @property
    def w(self):
        return self.__w
 
    @w.setter
    def w(self, value):
        if value <= 0:
            raise ValueError("Width must be positive")
        self.__w = value
 
    @property
    def h(self):
        return self.__h
 
    @h.setter
    def h(self, value):
        if value <= 0:
            raise ValueError("Height must be positive")
        self.__h = value
 
    def area(self):
        return self.__w * self.__h
 
    def perimeter(self):
        return 2 * (self.__w + self.__h)
 
    def translate(self, dx, dy):
        self.__p.translate(dx, dy)
 
    def __repr__(self):
        return f"Rectangle(bottom_left={self.__p}, width={self.__w}, height={self.__h})"
 
 
p1, p2, p3 = Point2d(1, 2), Point2d(3, 4), Point2d(5, 6)
r1, r2 = Rectangle(p1, 10, 20), Rectangle(p2, 5, 15)
 
print(p1, p2, p3)
print(r1, r2)
p1 += p3
print(p1, r1)
```

### Polymorphism

```python
class Animal:
    def __init__(self, name):
        print(f"{name}, is born!")
        self.name = name
        self.hungry = True
 
    def walk(self):
        self.hungry = True
        print(f"{self.name} is walking.")
 
    def eat(self):
        self.hungry = False
        print(f"{self.name} is eating.")
 
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed
 
    def walk(self):
        if self.hungry:
            print(f"{self.name} is too hungry to walk.")
        else:
            super().walk()
 
    def bark(self):
        print(f"{self.name} is barking.")
 
class Cat(Animal):
    def __init__(self, name, color):
        super().__init__(name)
        self.color = color
 
    def meow(self):
        print(f"{self.name} is meowing.")
 
c1 = Cat("Whiskers", "tabby")
d1 = Dog("Buddy", "Golden Retriever")
house = [c1, d1]
d1.eat()
for pet in house:
    pet.walk()
    pet.walk()
 
c1.meow()
d1.bark()
```

```python
# EX2: Implement Car and Truck classes.
# Both vehicles should have the methods:
# - drive(distance): drives the vehicle for a certain distance, consuming fuel based on the vehicle's consumption rate.
# - refuel(amount): adds fuel to the vehicle's tank, ensuring it does not exceed the fuel capacity.
# - status(): returns the current fuel level and odometer reading.
# Truck should have additional methods:
# - load_cargo(weight): loads cargo onto the truck.
# - unload_cargo(weight): unloads cargo from the truck.
# Bonus: if the truck is loaded the fuel consumption rate should increase by 20%.

class Vehicle:
    def __init__(self, consumption, fuel_capacity):
        self.__odometer = 0
        self.__fuel_level = fuel_capacity
        self.__consumption = consumption
        self.__fuel_capacity = fuel_capacity

    @property
    def odometer(self):
        return self.__odometer

    @odometer.setter
    def odometer(self, value):
        if value >= self.__odometer:
            self.__odometer = value
        else:
            raise ValueError("Odometer cannot decrease.")

    @property
    def fuel_level(self):
        return self.__fuel_level

    @property
    def consumption(self):
        return self.__consumption

    @property
    def fuel_capacity(self):
        return self.__fuel_capacity

    @fuel_level.setter
    def fuel_level(self, value):
        if 0 <= value <= self.__fuel_capacity:
            self.__fuel_level = value
        else:
            raise ValueError(f"Fuel level must be between 0 and {self.__fuel_capacity} liters.")

class Car(Vehicle):
    def drive(self, distance):
        fuel_needed = (self.consumption / 100) * distance
        if fuel_needed <= self.fuel_level:
            self.fuel_level -= fuel_needed
            self.odometer += distance
            return True
        else:
            max_distance = (self.fuel_level / self.consumption) * 100
            self.fuel_level = 0
            self.odometer += max_distance
            return False

    def refuel(self, amount):
        if amount < 0:
            raise ValueError("Refuel amount must be positive.")
        new_fuel_level = self.fuel_level + amount
        if new_fuel_level > self.fuel_capacity:
            raise ValueError(f"Cannot exceed fuel capacity of {self.fuel_capacity} liters.")
        self.fuel_level = new_fuel_level

    def status(self):
        return f"Fuel level: {self.fuel_level} liters, Odometer: {self.odometer} km"

class Truck(Vehicle):
    def __init__(self, consumption, fuel_capacity, cargo_capacity=1000):
        super().__init__(consumption, fuel_capacity)
        self.__cargo_weight = 0
        self.__cargo_capacity = cargo_capacity

    @property
    def cargo_weight(self):
        return self.__cargo_weight

    @property
    def cargo_capacity(self):
        return self.__cargo_capacity

    @cargo_weight.setter
    def cargo_weight(self, value):
        if value < 0:
            raise ValueError("Cargo weight cannot be negative.")
        if value > self.__cargo_capacity:
            raise ValueError(f"Cannot exceed cargo capacity of {self.__cargo_capacity} kg.")
        self.__cargo_weight = value

    def load_cargo(self, weight):
        if weight < 0:
            raise ValueError("Cargo weight to load must be positive.")
        self.cargo_weight = self.cargo_weight + weight

    def unload_cargo(self, weight):
        if weight < 0:
            raise ValueError("Cargo weight to unload must be positive.")
        self.cargo_weight = self.cargo_weight - weight

    def drive(self, distance):
        effective_consumption = self.consumption * (1.2 if self.cargo_weight > 0 else 1)
        fuel_needed = (effective_consumption / 100) * distance
        if fuel_needed <= self.fuel_level:
            self.fuel_level -= fuel_needed
            self.odometer += distance
            return True
        else:
            max_distance = (self.fuel_level / effective_consumption) * 100
            self.fuel_level = 0
            self.odometer += max_distance
            return False

    def refuel(self, amount):
        if amount < 0:
            raise ValueError("Refuel amount must be positive.")
        new_fuel_level = self.fuel_level + amount
        if new_fuel_level > self.fuel_capacity:
            raise ValueError(f"Cannot exceed fuel capacity of {self.fuel_capacity} liters.")
        self.fuel_level = new_fuel_level

    def status(self):
        return f"Fuel level: {self.fuel_level} liters, Odometer: {self.odometer} km, Cargo weight: {self.cargo_weight} kg"

# Tests

# Car tests:

car = Car(consumption=8, fuel_capacity=50)

print("Initial:", car.status())
car.drive(100)
print("After drive 100km:", car.status())

car.refuel(5)
print("After refuel 5L:", car.status())

# Car error cases:

# car.refuel(-1)
# car.refuel(10)
# car.fuel_level = 60

#Truck tests

truck = Truck(consumption=25, fuel_capacity=120, cargo_capacity=1000)

print('Initial:', truck.status())


truck.load_cargo(300)
print('After loading 300kg:', truck.status())

truck.unload_cargo(100)
print('After unloading 100kg:', truck.status())


# truck.load_cargo(1000)

# truck.unload_cargo(500)

# truck.load_cargo(-5)

# truck.unload_cargo(-5)
```
