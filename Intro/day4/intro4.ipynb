## Generators

```python
def gen_sq():
    num = 2
    while True:
        yield num ** 2
        num += 1
my_gen = gen_sq()
for _ in range(5):
    print(next(my_gen), end=" ")

```

## Fibonacci Series

```python
def gen_fib():
    a, b = 0, 1
    while True:
        yield b
        a, b = b, a + b

gen = gen_fib()
for _ in range(10):
    print(next(gen))
```

### Dictionary Comprehension

```python
nums = [1, 2, 3, 4, 5, 4, 3, 2, 2]
def dict_count(nums):
    count_dict = {}
    for num in nums:
        if num in count_dict:
            count_dict[num] += 1
        else:
            count_dict[num] = 1
    return count_dict

dict_count(nums)
```

```python
# EX2: Create a class called `BankAccount` that has the following attributes and methods:
# - Attributes:
#   - `account_number`: a string representing the account number
#   - `balance`: a float representing the account balance
# - Methods:
#   - `deposit(amount)`: a method that takes a float `amount` and adds it to the `balance`
#   - `withdraw(amount)`: a method that takes a float `amount` and subtracts it from the `balance` if there are sufficient funds, otherwise it should print an error message
#   - `get_balance()`: a method that returns the current balance
#   - `get_account_number()`: a method that returns the account number
 
class BankAccount:
    def __init__(self, account_number, balance=0.0):
        self.account_number = account_number
        self.balance = balance
 
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited ${amount:.2f}. New balance: ${self.balance:.2f}.")
        else:
            print("Deposit amount must be positive.")
 
    def withdraw(self, amount):
        if amount > 0:
            if self.balance >= amount:
                self.balance -= amount
                print(f"Withdrew ${amount:.2f}. New balance: ${self.balance:.2f}.")
            else:
                print("Insufficient funds for withdrawal.")
        else:
            print("Withdrawal amount must be positive.")
 
    def get_balance(self):
        return self.balance
 
    def get_account_number(self):
        return self.account_number

acc1 = BankAccount("204273569", 1000.0)
acc2 = BankAccount("240833608", 500.0)

acc1.withdraw(200.0)
acc2.deposit(300.0)
acc1.withdraw(1000.0)
acc2.deposit(-50.0)
```

```python
# Ex3: Create a class called `Car` that has the following attributes and methods:
# - Attributes:
#   - `consumption`: a float representing the fuel consumption in liters per 100 kilometers
#   - `fuel`: a float representing the current fuel level in liters
#   - `odometer`: a float representing the total distance driven in kilometers
# - Methods:
#   - `drive(distance)`: a method that takes a float `distance` in kilometers
#   - `refuel(amount)`: a method that takes a float `amount` in liters and adds it to the `fuel` level
#   - `status()`: a method that returns a string with the current fuel level and odometer reading

class Car:
    def __init__(self, consumption, fuel=0.0):
        self.consumption = consumption
        self.fuel = fuel
        self.odometer = 0.0
    def drive(self, distance):
        fuel_needed = (distance / 100) * self.consumption
        if self.fuel >= fuel_needed:
            self.fuel -= fuel_needed
            self.odometer += distance
            print(f"Drove {distance} km. Fuel left: {self.fuel:.2f} liters.")
        else:
            print("Insufficient fuel for this distance.")
    def refuel(self, amount):
        if amount > 0:
            self.fuel += amount
            print(f"Refueled {amount} liters. Current fuel: {self.fuel:.2f} liters.")
        else:
            print("Refuel amount must be positive.")
    def status(self):
        return f"Fuel: {self.fuel:.2f} liters, Odometer: {self.odometer:.2f} km."

car = Car(consumption=8.0, fuel=20.0)
car.drive(100)
car.refuel(10)
print(car.status())
```

```python
class Car:
    def __init__(self, maker, model):
        self.maker = maker
        self.model = model
 
    def move_to_spot(self, spot):
        print(f"{self.maker} {self.model} is moving to spot {spot}.")
 
    def exit_spot(self):
        print(f"{self.maker} {self.model} is exiting the parking spot.")
 
class Garage:
    def __init__(self):
        self.cars = []
 
    def park_car(self, car):
        self.cars.append(car)
        print(f"{car.maker} {car.model} has been parked in the garage.")
 
    def remove_car(self, car):
        if car in self.cars:
            self.cars.remove(car)
            print(f"{car.maker} {car.model} has been removed from the garage.")
        else:
            print(f"{car.maker} {car.model} is not in the garage.")
 
    def maintain_garage(self):
        print("Maintaining the garage and removing all cars.")
        for car in self.cars:
            car.exit_spot()
        self.cars.clear()
 
c1 = Car("Toyota", "Camry")
c2 = Car("Honda", "Civic")
c3 = Car("Ford", "Mustang")
garage = Garage()
garage.park_car(c1)
garage.park_car(c2)
garage.park_car(c3)
garage.maintain_garage()
```

```python
# add sortting feature to the point and rectangle classes
# they can be sorted by their location on the x-axis

class Point2d:
    def __init__(self, x, y):
        self.x = x
        self.y = y
 
    def __add__(self, other):
        if isinstance(other, Point2d):
            return Point2d(self.x + other.x, self.y + other.y)
        return NotImplemented
 
    def translate(self, dx, dy):
        self.x += dx
        self.y += dy
 
    def __repr__(self):
        return f"p({self.x}, {self.y})"
 
    def __lt__(self, other):
        return self.x < other.x
 
class Rectangle:
    def __init__(self, p, w, h):
        self.p = p  
        self.w = w  
        self.h = h  
 
    def area(self):
        return self.w * self.h
 
    def perimeter(self):
        return 2 * (self.w + self.h)
 
    def translate(self, dx, dy):
        self.p.translate(dx, dy)
 
    def __repr__(self):
        return f"r({self.p}, w:{self.w}, h:{self.h})"
 
    def __lt__(self, other):
        return self.p < other.p
 
 
points = [Point2d(3, 4), Point2d(1, 2), Point2d(5, 6)]
print("Points before sorting:", points)
points.sort()
print("Points after sorting:", points)
 
rectangles = [Rectangle(Point2d(3, 4), 5, 6), Rectangle(Point2d(1, 2), 7, 8), Rectangle(Point2d(5, 6), 9, 10)]
print("Rectangles before sorting:", rectangles)
rectangles.sort()
print("Rectangles after sorting:", rectangles)
```

```python
class Cat:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def meow(self):
        print(f"{self.name} says Meow!")

    def __lt__(self, other):
        return self.age < other.age

cats = [Cat("Cloud", 6), Cat("Marble", 9), Cat("Tres", 2)]
print("Cats:")
for cat in cats:
    print(f"{cat.name}, Age: {cat.age}")
if cats[0] < cats[1]:
    print(f"{cats[0].name} is younger than {cats[1].name}.")
else:
    print(f"{cats[0].name} is older than {cats[1].name}.")

cats.sort()
print("Cats sorted by age:")
for cat in cats:
    print(f"{cat.name}, Age: {cat.age}")
```

### OOP Excercise

```python
# oop composition
#
# Create a class Student with attributes name, surname, id
# with methods:
#   __init__(self, name, surname, id) -
#       takes name, surname and id as parameters
#   courses(self) -
#       returns a list of Course objects the student is enrolled in
#
# Create a class Course with attributes name, code, max_students
# with methods:
#   __init__(self, name, code, max_students) -
#       takes name, code and maximum number of students as parameters
#   enrolled_students(self) -
#       returns a list of Student objects enrolled in the course
#
# Create a class Enrollment with methods:
#   __init__(self, student, course) -
#       takes one Student object and one Course object
#   enroll(self) -
#       enrolls the student in the course if the maximum number
#       of students is not exceeded, returns True if successful,
#       False otherwise
#   unenroll(self) -
#       unenrolls the student from the course, returns True if
#       successful, False otherwise
# Create class University with methods that helps with statistics:
#   __init__(self) -
#       initializes an empty list of students, courses, enrollments, and
#       history of unenrollments
#   add_student(self, student) -
#       adds a Student object to the university
#   add_course(self, course) -
#       adds a Course object to the university
#   enroll_student(self, student, course) -
#       creates an Enrollment object and enrolls the student in the course
#   unenroll_student(self, student, course) -
#       unenrolls the student from the course
#   student_courses(self, student) -
#       returns a list of Course objects the student is enrolled in
#   course_students(self, course) -
#       returns a list of Student objects enrolled in the course
#   get_most_popular_course(self) -
#       returns the Course object with the highest number of enrolled students
#   get_student_with_most_courses(self) -
#       returns the Student object enrolled in the highest number of courses
#   get_course_with_fewest_students(self) -
#       returns the Course object with the lowest number of enrolled students
#   get_student_with_fewest_courses(self) -
#       returns the Student object enrolled in the lowest number of courses
#   get_unenrollment_history(self) -
#       returns a list of tuples enrollment records that were unenrolled
#   get_most_popular_historical_course(self) -
#       returns the Course object with the highest number of unenrollments

class Student:
    def __init__(self, name, surname, id):
        self.name = name
        self.surname = surname
        self.id = id
        self._courses = []

    def courses(self):
        return self._courses

class Course:
    def __init__(self, name, code, max_students):
        self.name = name
        self.code = code
        self.max_students = max_students
        self._enrolled_students = []

    def enrolled_students(self):
        return self._enrolled_students

class Enrollment:
    def __init__(self, student, course):
        self.student = student
        self.course = course

    def enroll(self):
        if len(self.course.enrolled_students()) < self.course.max_students:
            self.course.enrolled_students().append(self.student)
            self.student.courses().append(self.course)
            return True
        return False

    def unenroll(self):
        if self.student in self.course.enrolled_students():
            self.course.enrolled_students().remove(self.student)
            self.student.courses().remove(self.course)
            return True
        return False

class University:
    def __init__(self):
        self.students = []
        self.courses = []
        self.enrollments = []
        self.unenrollment_history = []

    def add_student(self, student):
        self.students.append(student)

    def add_course(self, course):
        self.courses.append(course)

    def enroll_student(self, student, course):
        enrollment = Enrollment(student, course)
        if enrollment.enroll():
            self.enrollments.append(enrollment)
            return True
        return False

    def unenroll_student(self, student, course):
        enrollment = next((e for e in self.enrollments if e.student == student and e.course == course), None)
        if enrollment and enrollment.unenroll():
            self.enrollments.remove(enrollment)
            self.unenrollment_history.append((student, course))
            return True
        return False

    def student_courses(self, student):
        return student.courses()

    def course_students(self, course):
        return course.enrolled_students()

    def get_most_popular_course(self):
        return max(self.courses, key=lambda c: len(c.enrolled_students()), default=None)

    def get_student_with_most_courses(self):
        return max(self.students, key=lambda s: len(s.courses()), default=None)

    def get_course_with_fewest_students(self):
        return min(self.courses, key=lambda c: len(c.enrolled_students()), default=None)

    def get_student_with_fewest_courses(self):
        return min(self.students, key=lambda s: len(s.courses()), default=None)

    def get_unenrollment_history(self):
        return self.unenrollment_history

    def get_most_popular_historical_course(self):
        from collections import Counter
        if not self.unenrollment_history:
            return None
        course_counts = Counter(course for student, course in self.unenrollment_history)
        most_common_course, _ = course_counts.most_common(1)[0]
        return most_common_course



```
