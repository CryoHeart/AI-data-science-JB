### LuxuryCar

Create a class called `LuxuryCar` that inherits from `Car`.

Every rental has an additional fixed cleaning fee of `100 NIS`.

Override `calculate_price(days)` and use `super()` to reuse the existing price calculation.

Example:

```text
Mercedes S-Class
700 NIS/day
2 days

700 × 2 + 100 = 1500 NIS
```


````python
class Vehicle:
    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price
        self.is_available = True

    def get_description(self):
        print(f"{self.brand} {self.model}. {self.price} per day.")
        if (self.is_available):
            print("Available")
        else:
            print("Not available.")

    def calculate_price(self, days):
        price_for_days = self.price * days
        print(f"The price for {days} days will be {price_for_days}.")

    def rent_car(self, days):
        self.is_available = False
        full_car_name = f"{self.brand} {self.model}"
        if (days < 0):
            return "Dr Emmet Brown, is that you?"
        else:
            print(f"{full_car_name} was rented for {days} days.")

    def return_car(self):
        self.is_available = True
        full_car_name = f"{self.brand} {self.model}"
        print(f"{full_car_name} has been returned.")

skoda_fabia = Vehicle("Skoda", "Fabia", 200)

# skoda_fabia.get_description()
# skoda_fabia.calculate_price(4)
# skoda_fabia.rent_car(5)
# skoda_fabia.get_description()
# skoda_fabia.return_car()
# skoda_fabia.get_description()

class Car(Vehicle):
    def __init__(self, brand, model, price, number_of_seats):
        super().__init__(brand, model, price)
        self.number_of_seats = number_of_seats

    def get_description(self):
            print(f"{self.brand} {self.model}. {self.number_of_seats} seats. {self.price} per day.")
            if (self.is_available):
                print("Available")
            else:
                print("Not available.")

nissan_micra = Car("Nissan", "Micra", 150, 5)

# nissan_micra.get_description()
# nissan_micra.calculate_price(6)

class ElectricCar(Car):
    def __init__(self, brand, model, price, number_of_seats):
        super().__init__(brand, model, price, number_of_seats)
        self.battery_level = 100

    def drive(self, km):
        if (km < 0):
            return "This isn't Dr. Brown's DeLorian..."
        
        self.battery_level = self.battery_level - (km * 0.5)
        full_car_name = f"{self.brand} {self.model}"
        if (self.battery_level > 0):
            print(f"{full_car_name} drove for {km} km. {self.battery_level}% battery remains.")
        if (self.battery_level < 20):
            print("Battery low. Please recharge.")
        elif (self.battery_level <= 0):
            return "Battery depleted. Please recharge."

    def get_description(self):
        super().get_description()
        service_fee = (self.price // 10)
        full_price = service_fee + self.price
        print(f"{self.brand} {self.model}. {self.number_of_seats} seats.")
        print(f"Rental price (Per day): {self.price}ILS. Service fee: {service_fee}ILS. Total price per day: {full_price}ILS")
        self.price = full_price

# tesla_model_6 = ElectricCar("Tesla", "Model 6", 300, 5)
# tesla_model_6.get_description()
# tesla_model_6.drive(10)

class LuxuryCar(Car):
    def __init__(self, brand, model, price, number_of_seats):
        super().__init__(brand, model, price, number_of_seats)
        self.cleaning_fee = 200


    def calculate_price(self, days):
        super().calculate_price(days)
        print(f"{self.brand} {self.model}")
        print(f"{self.price}ILS/day")
        print(f"{self.price} * {days} + {self.cleaning_fee} = {self.price * days + self.cleaning_fee}ILS.")


nissan_GTR = LuxuryCar("Nissan", "GTR", 700, 4)
nissan_GTR.calculate_price(5)
````

## Bonus

Extend `sort_songs()` so that it can sort by more than one field.

For example:

```text

```

Songs should first be grouped by artist and then sorted by year within each artist.


````python
class Song:
    def __init__(self, title, artist, duration, year):
        self.title = title
        self.artist = artist
        self.duration = duration
        self.year = year

    def getInfo(self):
        print(f"{self.title} - {self.artist} ({self.year}). Duration: {self.duration} sec.")

bohemian_rhapsody = Song("Bohemian Rhapsody", "Queen", 354, 1975)
show_must_go_on = Song("Show Must Go On", "Queen", 270, 1991)
another_one_bites_the_dust = Song("Another One Bites The Dust", "Queen", 215, 1980)
love_of_my_life = Song("Love of My Life", "Queen", 220, 1975)
bicycle_race = Song("Bicycle Race", "Queen", 180, 1978)

class Playlist:
    def __init__(self, name, collection_of_songs):
        self.name = name
        self.collection_of_songs = list(collection_of_songs)


    def add_song(self, song):
        self.collection_of_songs.append(song)

    def remove_song(self, song_title):
        self.collection_of_songs = [song for song in self.collection_of_songs if song.title != song_title]

    def get_total_duration(self):
        total_duration = sum(song.duration for song in self.collection_of_songs)
        print(f"Total duration of playlist '{self.name}': {total_duration} sec.")

    def show_songs(self):
        print(f"Playlist: {self.name}")
        for song in self.collection_of_songs:
            song.getInfo()

    def sort_song(self, method):
        if method == "title":
            self.collection_of_songs.sort(key=lambda song: song.title)
        elif method == "artist":
            self.collection_of_songs.sort(key=lambda song: song.artist)
        elif method == "year":
            self.collection_of_songs.sort(key=lambda song: song.year)
        elif method == "duration":
            self.collection_of_songs.sort(key=lambda song: song.duration)
        else:
            print("Invalid sorting method.")

    def find_song(self, keyword):
        if isinstance(keyword, int):
            found_songs = [song for song in self.collection_of_songs if song.year == keyword]
        else:
            found_songs = [song for song in self.collection_of_songs if keyword.lower() in song.title.lower() or keyword.lower() in song.artist.lower()]
        if found_songs:
            print(f"Songs found with keyword '{keyword}':")
            for song in found_songs:
                song.getInfo()
        else:
            print(f"No songs found with keyword '{keyword}'.")

queen_playlist = Playlist("Queen", [bohemian_rhapsody])

queen_playlist.add_song(show_must_go_on)
queen_playlist.add_song(another_one_bites_the_dust)
queen_playlist.add_song(love_of_my_life)
queen_playlist.add_song(bicycle_race)

queen_playlist.sort_song("title")
queen_playlist.show_songs()

queen_playlist.find_song(1991)

class PremiumPlaylist(Playlist):
    def __init__(self, name, collection_of_songs):
        super().__init__(name, collection_of_songs)
        self.max_duration = 400

    def add_song(self, song):
            super().add_song(song)
            if (song.duration > self.max_duration):
                print("Song exceeds playlist's duration limit.")
            else:
                self.collection_of_songs.append(song)
                print(f"{song.title} - {song.artist} has been added to the playlist {self.name}.")

class PodcastEpisode:
    def __init__(self, title, host, duration, episode_number):
        self.title = title
        self.host = host
        self.duration = duration
        self.episode_number = episode_number

    def getInfo(self):
        print(f"{self.title} - {self.host} (Episode {self.episode_number}). Duration: {self.duration} sec.")

````
